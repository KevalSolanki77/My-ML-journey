```python
import numpy as np

import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering
```

```python
# Generate synthetic data
X, y = make_blobs(n_samples=100, centers=3, cluster_std=0.8, random_state=0)
```

```python
plt.title("Dendogram")
dend = dendrogram(linkage(X, method='ward'))
```

```python
# Fit with a fixed number of clusters to get sample labels
cluster_model = AgglomerativeClustering(
    n_clusters=3, metric='euclidean', linkage='ward'
)
labels = cluster_model.fit_predict(X)
```

```python
# 6. Visualize the clustered 2D points
plt.figure(figsize=(6, 5))
plt.scatter(
    X[:, 0], X[:, 1], c=labels, cmap='viridis', s=60, edgecolors='black'
)
plt.title('Agglomerative Clustering Output (k=3)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
